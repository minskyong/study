# 항만 병목 요인과 분석 프레임

항만도 선박이 오래 기다리거나 컨테이너가 늦게 나가게 되면 병목현상이 발생할 수 있다. 항만에서의 병목 현상은 항만 전체가 아니라 어떤 흐름 중 한 단계에 작업이 몰리거나 지연이 있어서 발생할 수 있다.

항만에서 병목을 만드는 요인을 해측/터미널측/육상측 정보 영역으로 나누어 정리하고, 공개데이터를 이용하여 병목을 측정하기 위한 선석/대기/체류 분석 프레임을 살펴보겠다.

## 병목현상

병목(Bottleneck)은 여러 단계가 이어진 작업 흐름에서 처리 능력이 가장 작기 때문에 전체 흐름의 속도를 결정할 수 있는 단계를 의미한다.

병목이 아닌 단계를 아무리 개선해도 전체속도에는 영향이 없다. 병목을 해소한다고 해서 모든 문제가 해소되지는 않는다. 그 다음으로 느렸던 단계가 새로운 병목이 될 수 있기 때문이다.

그렇기 때문에 병목 분석은 한 번의 진단으로 끝내는 것이 아니라 개선후에 다시 측정하는 반복과정이 있어야된다.

### 처리능력과 수요

병목을 측정하고 관리하기 위해서는, 병목을 숫자로 살펴볼 수 있어야 한다. 각각의 단계에 대해서 처리능력(Capacity), 해당 단계에 들어오는 수요(Demand)를 비교해야 한다.

처리능력은 한 단계가 정해진 시간 안에 최대로 처리할 수 있는 양, 수요는 실제로 그 단계에 도착한 양을 의미한다.

수요가 처리 능력보다 크지 못하면 처리하지 못한양이 대기열이 되고, 그 대기열이 앞 단계까지 거꾸로 밀려 올라가기 때문이다.

이 두 값의 비율을 이용율이라고 일컫는다. 선석의 경우에는 선박으로 사용되고있던 시간의 비율로 볼수있고 이것이 바로 선석 점유율로 볼 수 있다.

- 이용률 = 수요 / 처리 능력 * 100
- 선석점유율 = 선박이 선석에 머문 시간의 총합 / 관측 기간의 전체시간 * 100



## 항만 병목 요인

항만에서 병목현상이 발생할 수 있는 요인이 무엇인가?

- 선석이 부족해서 병목현상이 발생할 수 있다.
- 선석은 충분하지만 장치장이 가득차서 하역이 느려졌다.
- 트럭이 한꺼번에 몰려서 게이트가 막혀서 병목현상이 발생할 수 있다.

- 해측: 선석 부족, 선박 대형화, 도착 불규칙, 도선/예선의 부족, 기상
- 터미널 측: 하역 장비 생산성, 장치장 적재율, 작업/인력 교대
- 육상 측: 게이트 처리능력, 트럭 반출입 시간 몰림, 배후 도로/철도 관리상문제
- 정보/제도: 신고/허가 지연, 기관간 정보공유부족, 명절(기간특수성)

## 분석 프레임

분석을 시작하기 이전에 무엇을 어떤 지표로, 어떤 데이터로 볼 것인가를 결정해두어야 하는데 이를 분석 프레임(Analysis Framework)라고 일컫는다.

건강검진시 혈압, 혈당, 콜레스트롤 수치 등 정해진 항목을 늘 같은 방식으로 측정한다. 분석 프레임도 마찬가지의 개념, 항목과 측정 방법을 고정해두어야 시기별/부두별 결과를 서로 비교할 수 있기 때문에 미리 정해두는것을 의미한다.

선박 한 척이 항만에서 겪는 시간을 세부분(선석, 대기, 체류)으로 나누어 프레임을 결정한다.

- 선석: 시설이 얼마나 사용중인지(계선장소, 시설코드)
- 대기: 선박이 부두에 붙기 전에 얼마나 기다렸나(선박입항신고, 관제정보)
- 체류: 선박이 항만에 머문 시간이 계획적이었나(관제정보, 선박입항신고)



### 지표 계산식

계산식을 미리 만들어두면 '누가 계산해도 같은 결과 얻을 수 있고, 제외 조건을 누락시키는 실수도 줄일 수 있다'.

- 예정 체류시간 = (출항예정일시 - 입항일자) / 시간
- 선석 점유율 = 선석에 선박이 한 척 있었던 시간 / 관측 기간 전체 시간 * 100
- 평균 동시 척수(선석에 동시에 있는 배의 갯수) = 선석별 예정 체류시간의 합 / 관측 기간 전체 시간
- 묘박지 입항 비율 = 계선장소가 묘박지(수역시설)인 입항건수 / 전체 입항 건수 * 100
- 실제 체류시간 = (출항 일시 - 입항일시) / 시간
- 출항지연시간 = 출항 일시 - 출항 예정일시

## 분석 절차

1. 질문 정하기: 어느 관점(선석/대기/체류)의 어떤 병목을 볼것인가
2. 데이터 선택: 어떤 데이터가 필요한가
3. 품질 확인: 선택된 데이터의 품질은 어떠한가
4. 지표 계산: 프레임의 계산식과 제외조건을 적용한 지표를 계산
5. 비교: 선석, 부두, 월별, 선종간 비교
6. 원인 가설: 데이터로 보이는 증상과 보이지 않는 원인을 나누어 정리


병목의 성질

- 병목이 아닌 단계를 개선해도 처리량이 늘거나 하지는 않는다.
- 병목을 해소하면 그 다음으로 느린 단계가 다시 병목이 된다.

수요가 병목의 처리능력보다 크면 처리하지 못하는 양이 매일 누적되는데, 이를 적체라고 일컫는다.

## 이용률, 선석 점유율

병목을 찾으려면 단계마다 얼마나 바쁜가를 기준으로 비교해야한다. 이용률이 사용된다.

항만에서는 선석의 이용률을 선석점유율로 나타내서 계산한다.

- 선석 점유율 = 선박이 선석에 머문 시간의 합 / 관측 기간 전체 시간

한달 = (선박 22척이 평균 26시간을 머무름) 572 / 720 * 100

## 이용률과 대기시간

점유율 80%, 20%

- p(이용률): 0~1 사이로 나타낸 값
- 평균 작업시간: 선박 한척이 선석을 쓰는 평균 시간

```
평균 대기시간 = p / ( 1-p) * 평균 작업시간
```

작업시간이 24시간인 선석에서 이용률이 50%면 평균 대기는 24시간이지만, 80%면 96시간, 90%면 216시간으로 이용률이 조금 오를때마다 대기시간이 곱절로 증가하게 된다.

이용률이 1에 가까워지면 분모는 (1-p)에 가까워지기떄문에 대기가 끝없이 커진다.

- 선석이 하나이고, 도착과 작업시간이 모두 무작위

````python
period_start, period_end = pd.Timestamp('2024-05-01'), pd.Timestamp('2025-08-01')
period_hours = (period_end - period_start).total_seconds() / 3600

period_hours
````

````python
def occupied_hours(starts, ends):
    # 체류구간들을 시작 시각 순으로 이어 붙인다음에 겹침없이 선박이 있던 총 시간을 구함
    interval = sorted(zip(starts, ends))
    total = pd.Timedelta(0)
    current_start, current_end = interval[0]
    for start, end in interval[1:]:
        if start <= current_end:
            current_end = max(current_end, end)
        else:
            total += current_end - current_start
            current_start, current_end = start, end
    total += current_end - current_start
    return total.total_seconds() / 3600
````

````python
rows = []

for code, group in berth_calls.groupby('계선장소시설코드'):
    rows.append({
        '코드': code,
        '선석': group['계선장소명'].mode()[0],
        '시설구분': group['시설구분'].iloc[0],
        '입항건수': len(group),
        '점유율': occupied_hours(group['입항'], group['출항예정']) / period_hours * 100,
        '평균동시척수': group['예정체류'].sum() / period_hours,
    })

berth_index = pd.DataFrame(rows).set_index('코드')
berth_index.shape
````
