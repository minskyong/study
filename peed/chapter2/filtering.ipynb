# Series의 이해 
Series에 저장된 여러 값을 범위로 선택  <br>
조건에 맞는 값만 추출해 값을 수정하거나 연산하는 방법

```python
import pandas as pd
import numpy as np
```

```python
sales = pd.Series(
        [100, 200, 300, 400 , 500],
        index = ['월','화','수','목','금'],
        name = '일별 판매량',
)
sales
```

```python
# 슬라이싱을 해야할 떄  .iloc사용 

sales.iloc[1:4]
sales.iloc[3:100]   #예외사항 : 슬라이싱할 때 인덱스 넘어도 오류x  --> shape 통해서 갯수 확인
                    #print(f'shape = {sales.shape}')
```

```python
sales.iloc[10:20]      #새로운 시리즈를 가져오는 것
```

```python
sales.loc['화':'목']   #인덱스로 가져올 수  있음.
```

```python
#월 수 금을 뽑아 오고 싶을 때

sales.loc['월']
sales.loc['수']
sales.loc['금']

a= ['월', '화', '수']              #for문을 사용
for date in a:
    print(sales.loc[date])
```

```python
sales.loc[['월','화','수']]   #loc는 배열로 묶어서 특정 인덱스 한번에 가져오기 가능 
```
