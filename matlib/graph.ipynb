```python
product = [ 'product a', 'product b', 'product c', 'prouct d']

sales = [120, 180, 210, 260]

fig , ax =plt.subplots(figsize = (10,4))

bars = ax.barh(product, sales)

ax.barh(product, sales)  

ax.bar_label(bars, padding = 3)

ax.set_title ('product sales')
ax.set_xlabel('sales')

plt.show()
```

```python
product = ['a','b','c']

sales_2026 = [120,140,160]
sales_2025 = [110,200, 90]

x = np.arange(len(product))
width = 0.35

w= width 

fig, ax = plt.subplots(figsize=(8,4))

ax.bar (

    x- w /2 ,
    sales_2025,
    w,
    label = '2025',
)
ax.bar (
    x+ w /2 ,
    sales_2026,
    w,
    label = '2026')

ax.set_xticks(x)
ax.set_xticklabels(product)
ax.set_title('productsales')
ax.set_xlabel('product')
ax.set_ylabel('sales')
ax.legend()

plt.show()
```

```python
# 누적 막대 그래프

product = ['a','b','c']

online = [120,140,160]
offline = [110,200, 90]

fig ,ax = plt.subplots(figsize = (10,4))


ax.bar(
    product,
    online,
    label = 'online'
)

ax.bar(
    product,
    offline,
    bottom = online,
    label = 'offline'
)

ax.bar(                     #추가 
    product,
    [10,20,30],
    label = 'others'
)
ax.legend()




plt.show()
```

```python

df = pd.DataFrame(
    {"study_hours": [1,2,3,4,5,5.5,6,6.5],
     "scores" : [60,70,65,75,80,82,83,90],
     'class' :['a','a','a','a','b','b','b','b']
     
     })

fig, ax = plt.subplots(figsize = (8,6))


class_a = df[df['class'] == 'a']
class_b = df[df['class'] == 'b']

ax.scatter(
 class_a['study_hours'],
 class_a['scores'],
 label = 'class a'


)

ax.scatter(
 class_b['study_hours'],
 class_b['scores'],
 label = 'class b'


)

ax.set_xlabel('study_hours')
ax.set_ylabel('score')

ax.legend()
```

```python
rng = np.random.default_rng(42)

x= rng.normal(
    loc = 50,
    scale =10,
    size = 500
)

y = ( x * 0.7 + rng.normal(loc = 0, scale = 8, size = 500))

print(len(x),len(y))

fig, ax =  plt.subplots(figsize = (8,6))

ax.scatter(x,y, s= 40, alpha =0.3)

plt.show()
```
