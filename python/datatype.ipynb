# 연산자

계산을 하기 위한 기호 
연산자(Operator): 특정한 계산이나 처리를 수행하도록 지시하는 기호 또는 키워드

연산자에 의해 처리되는 대상이 되는 값이나 변수를 피연산자(oeperand)라고 함


ex. 1+1 
1 = 피연산자 + = 연산자 

## 연산자 구분
연산자는 자신이 처리하는 피연산자 갯수에 따라 구분이 가능함 

### 피연산자 갯수에 따른 구분 

#### 단항연산자( unary operator)
단항연산자는 하나의 피연산자에 대해 연산을 수행하는 연산자들을 의미
Ex. sqrt? ++ -- 

#### 이항연산자 (binary operator)
두 개의 피연산에 대해 연산을 수행하는 연산자를 의미 
Ex. 사칙연산 

#### 다항연산자(ternary operator) = 삼항연산자
피연산자를 세게 혹은 그 이상을 요구하는 연산자를 의미 

### 기능에 따른 구분

#### 산술 연산자(arithmetic operator)
수학적 계산을 수행하는데 사용되는 연산자
한 개 or 두 개의 숫자의 수학적인 작업을 처리하는데 사용



```python
#파이썬은 정수와 실수 합산 가능 

a,b,c,d,e = 1,2,3,4.1,True
a_result = b+d
b_result = d+e
print(a_result)
print(b_result)

#문자열 합산도 가능

hello_str = "hello"
world_str = "world"

result =(hello_str + world_str)
print(result)

#배열 합산도 가능 
list1 = [1,2,3]
list2 = [3,4,5]

list_result = list1 + list2
print(list_result)
```

```python
#* = 곱 ** 제곱
# / = 몫.나머지 표기 //=몫만 표기
# %:모듈러 연산자

print(4**10)
print(10/4)
print(10//4)
```
