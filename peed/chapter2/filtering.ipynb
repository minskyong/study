```python
import pandas as pd
import numpy as np
```

```python
sales = pd.Series(
        [100, 200, 300, 400 , 500],
        index = ['월','화','수','목','금'],
        name = '일별 판매량',
)
sales
```

```python
# 슬라이싱을 해야할 떄  .iloc사용 

sales.iloc[1:4]
sales.iloc[3:5]   #예외사항 : 슬라이싱할 때 인덱스 넘어도 오류x  --> shape 통해서 갯수 확인
                    #print(f'shape = {sales.shape}')
```

```python
sales.iloc[10:20]      #새로운 시리즈를 가져오는 것
```

```python
sales.loc['목':'금']   #인덱스로 가져올 수  있음.
```

```python
#월 수 금을 뽑아 오고 싶을 때

sales.loc['월']
sales.loc['수']
sales.loc['금']

a= ['월', '화', '수']              #for문을 사용
for date in a:
    print(sales.loc[date])
```

```python
sales.loc[['월','화','수']]   #loc는 배열로 묶어서 특정 인덱스 한번에 가져오기 가능 
```

# Series 조건 필터링 

조건에 맞는 데이터들만 필터링해서 가져오는 방법    

```python
import pandas as pd
import numpy as np



scores = pd.Series(
            [80,90,70,60,50,55,65],
            index = ['학생1','학생2','학생3','학생4','학생5','학생6','학생7'],
            name = '중간고사 점수'




)
scores


```

## 조건 필터링(boolean series)
pandas 라이브러리 series는 비교 연산자의 연산을 지원 <br>
ary[] > 수 가능



```python
scores > 15
```

```python
# scores[scores > 80  and < 90 사용 불가 --> and는 논리 연산자 논리 자료형에서만 사용 가능


(scores >80) & (scores < 90)  # && , ||, ! and or not --> 비트논리연산자로 사용 가능 &, |,~
(scores >80) | (scores < 90)
~(scores > 80) | (scores < 90)
```

```python
scores.loc[(scores >=80) & (scores < 90)]    #긴거 보기 싫다하면 아래처럼

test = (scores >= 80) & (scores < 90)
scores.loc[test]
```

```python
~(scores > 90)   #not 
```

## between , isin
'''
(scores >=80) & (scores < 90)
'''

```python
scores.between(74,88)    #불린형식으로 나옴 수로 표현 하고 싶다면 loc사용
scores.loc[scores.between(74,88)]
```

```python
scores.loc[scores.between(74,88, inclusive = 'left')]   #왼쪽 삭제
scores.loc[scores.between(74,88, inclusive = 'right')]  #오른쪽 삭제
scores.loc[scores.between(74,88, inclusive = 'neither')]  #좌우 삭제 
```

```python
nums =pd.Series(
        [1,2,3,4,5]
)

nums.isin([2,3,4,5])    #불린 형식
nums.loc[nums.isin([2,3,4,5,])]
```

```python
#문자열 필터링

fruits = pd.Series(['apple','banna','cherry','durian','echo'])

fruits
```

```python
fruits.str.startswith('a')   #.str.startswith(" ") 사용시 " "로 시작하는거 불린형식 표현
fruits.loc[fruits.str.startswith('a')]  #.loc 사용해 마스킹으로 원하는 값 표현
```

## boolean series 이용한 집계
python은 논리 자료형을 내부적으로 정수 취급해 연산하는 것이 가능
- TRUE = 1
- FALSE = 0

```python
scores = pd.Series(
            [80,90,70,60,50,55,65],
            index = ['학생1','학생2','학생3','학생4','학생5','학생6','학생7'],
            name = '중간고사 점수'




)
```

```python
# 80점 이상인 사람 찾기

print(f'합격자 수 = {(scores >= 80).sum()}')   #scores >= 80 True,False 나온거 .sum()으로 뽑기

print(f'합격률 = {(scores >=80).mean()}')   #.mean 평균

print(f'만점자 = {(scores == 100).any()}')   #.any 하나라도

print(f'전원이 60점 이상인가? = {(scores >=60).all()}') #.all 전원
```
