```python
# 색상 종류

# red yellow blue green orange black 등 여러개 

```

```python
x = [1,2,3,4,5]
a = [10,15,20,25,30]
b = [8,13,18,23,28]
c = [6,11,16,21,26]


plt.plot(x,a, color = 'red', marker = 'o',label = 'A')
plt.plot(x,b, color = 'blue', marker = 'o', label = 'B')
plt.plot(x,c, color = 'yellow',marker = 'o',  label = 'C')
plt.legend()
plt.show()
```
