# 이동평균

이동평균(Moving Average)은 시계열에서 연속된 몇 개의 값을 묶어서 평균을 내고, 그 묶음을 한칸씩 옮기며 반복해 만드는 새로운 시계열을 의미한다. 한 번에 묶는 값은 윈도우(Window)라고 일컫는다.

파도가 치는 바닷가에서 물높이를 재면 파도 하나하나에 따라서 값이 크게 요동친다. 최근 몇번의 측정값을 평균내면 요동치는 값은 상쇄되고, 밀물, 썰물과 같은 큰 흐름만 남게 된다.

이동평균은 가파른 움직임을 걸러내고 느린 흐름만 남기는 필터로 정리할 수 있다.

- k개월 후행 이동평균: MA_k(t) = (y(t) + y(t-1) + ... + y(t-k+1)) / k

## 이동평균의 종류

- 후행 이동평균(Trailing): 윈도우의 마지막시점에 평균을 붙인다.
- 중심 이동평균(Centered): 윈도우의 가운데시점에 평균을 붙인다.
- 지수 이동평균(Exponential): 마지막 시점, 최근 값에 더 큰 무게

### 중심 이동평균

후행 이동평균은 이 시점의 추세가 얼마를 알고싶은가를 제대로 표현하기가 어렵다. 1~3월의 평균 2월 무렵의 수준을 나타내는데, 3월자리에 적게되면 추세선이 한달 늦게 그려지게된다.

평균을 가운데시점에 붙이게되면 이 부분을 해소할 수 있게된다. 중심 이동평균이라고 일컫는다. `rolling(n, center=True)`

중심 이동평균은 미래값을 써서 함께 계산을 하기때문에, 과거 자료를 분석할 때 주로 사용하고, 실시간 모니터링 예측에는 사용하기 어렵다. 앞뒤값 NaN이 발생할 가능성이 있기 때문에.

## 단순 이동평균

단순 이동평균은 윈도우 안의 값을 모두 더해서 윈도우 크기로 나눈 값을 의미한다. 3개월 이동평균이라면, 1~3월 값의 평균을 3월자리에, 2~4월 평균을 4월자리에 위치시키고 한칸씩 옮겨가는 방식을 의미한다.

`.rolling(n).mean()`: n개월 윈도우 계산할때 평균(mean)으로 계산한다. 합계(sum), 최대값(max)같은 연산도 가능하다.

3개월 이동평균 = (값(t-2) + 값(t-1) + 값(t)) / 3


### 윈도우 크기, 지연(Lag)

윈도우 크기를 키우면, 선이 매끄러워진다. 실제 반영을 다소 늦게 반영하는 현상이 발생한다. 지연이라고 일컫는다.

실제 물동량이 어느 달부터 갑자기 한단계 늘었다고 하더라도, 이동평균은 윈도우 안의 예전 값이 남아있는 동안 천천히 올라가기 때문에 실제 변경내용을 바로 반영하는것이 어렵기 때문이다.

윈도우 크기는 '무엇을 걷어내고 싶은가'로 결정해야된다. 한두달짜리 변동사항을 줄이고 최근 방향 전환을 빨리보고싶으면 3개월이 적당하고, 계절 패턴을 완전히 걷어내고 연간 흐름을 보고싶다라면은 12개월로 설정한다.
