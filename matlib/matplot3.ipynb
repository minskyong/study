```python


fig , ax = plt.subplots(
    figsize = (8,4)
        )
x= [1,2,3,4,5]
y = [10,15,20,25,30]

ax.plot(x,y,
        linestyle = 'dashed',  #점선 
        linewidth = 4,
        marker = 'o',  #마커 변경 
        markersize = 15
        )

```

```python
# 한 박스 안에 여러 그래프 넣기

fig , ax = plt.subplots(figsize = (8,4))

month = [1,2,3,4,5,]
product_a = [30,40,50,60,70]
product_b = [100,110,121,133,144]

ax.plot(month,product_a, marker = 'o', label = 'product_a')
ax.plot(month,product_b, marker = 's', label = 'product_b')

ax.legend()

plt.show()

```
