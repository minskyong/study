# 제어문 (control flow statement)
명령의 흐름을 제어하는데 사용하는 문법



## 조건문 (conditional statement)
주어진 조건에 따라 프로그램이 서로 다른 코드 블록을 실행하도록
분기시키는 제어구조 

## if
가장 기본적인 선택문
if문 다음 오는 값이나 조건식의 결과가 참일때만 해당
코드블럭의 명령문을 실행하는 방식으로 동작

구조
""" python
 if 값 or 조건식(절):
     명령문
"""


기본 if문 구조

if 값이나 조건식 :
    명령문 

```python
temperature = 35

if temperature > 30:
    print("덥다")
    print("물 먹어라")

print("현재 기온 :" , temperature )
```
