```python
import pandas as pd
import numpy as np
```

```python
#문제 1

scores = pd.Series([85, 92, 78, 88],
            index =['학생1', '학생2','학생3','학생4'])

print(scores)
```

```python
#문제 2
print(scores.to_list()) #값
print(scores.index) #인덱스 목록
print(type(scores)) #타입
len(scores) #크기
print(scores.shape) #형태
print(scores.ndim) #차원
print(scores.dtype) #타입
```

```python
# 문제 3

student = {                             #딕셔너리화

            "학생1" : 85 ,               #키 : 값 입력
            "학생2" : 92,
            "학생3" : 78,
            "학생4" : 88
}

s= student                             #긴 단어 줄이기

test = pd.Series(s, index=["학생1","학생2","학생3","학생4"])  # 변수 이름 test 설정
print("학생3의 성적은",test["학생3"],"이다")  #출력 
 
```

```python
#문제 4
series = pd.Series(                     #인덱스 A b C 네임텍 PRICE 100 200 300 
    [100, 200, 300],
    index=["A", "B", "C"],
    name="price"
)

print(series.index.to_list())         # A B C      인덱스 리스트
print(series.to_list())               # 100 200 300  
print(series.name)                    # price
print(series.size)                    # 3 ->시리즈 값 표현 
print(series.shape)                   #(3,) 시리즈 모양 표현
print(series.ndim)                    #1 차원
```
