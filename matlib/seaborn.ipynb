## long-form 데이터
하나의 열이 하나의 변수, 하나의 행이 하나의 관측값 갖는 구조

```python
long_sales = pd.DataFrame({
    "month": [
        "Jan", "Jan", "Jan",
        "Feb", "Feb", "Feb",
        "Mar", "Mar", "Mar",
        "Apr", "Apr", "Apr",
    ],
    "region": [
        "Seoul", "Busan", "Daejeon",
        "Seoul", "Busan", "Daejeon",
        "Seoul", "Busan", "Daejeon",
        "Seoul", "Busan", "Daejeon",
    ],
    "sales": [
        120, 100, 90,
        150, 130, 115,
        180, 160, 145,
        210, 190, 175,
    ],
})

long_sales
```

```python
sns.lineplot(

data = long_sales,
 x= 'month',
 y = 'sales',
 hue = 'region'
)
```

# wide-form 데이터
하나의 측정 변수가 여러 열로 펼쳐져 있는 형태 


```python
wide_sales = pd.DataFrame(
    {
        "Seoul": [120, 150, 180, 210],
        "Busan": [100, 130, 160, 190],
        "Daejeon": [90, 115, 145, 175],
    },
    index=["Jan", "Feb", "Mar", "Apr"],
)

wide_sales.index.name = "month"

wide_sales
```

```python
sns.lineplot(data =wide_sales)
plt.show()
```
