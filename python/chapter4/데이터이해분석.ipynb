# 데이터의 이해와 데이터 분석

## 데이터 분석

### 문제를 정의하고 관련 데이터를 수집 정리 탐색하여 질문에 대한 근거를 찾는 과정

- 좋은 분석은 계산이나 그래프보다 정확한 질문 생성
- 자료가 보여주는 의미를 현실 맥락에서 해석
- 복잡한 모델보다 의사결정에 필요한 비교와 집계가 가치 가능성 up


## 데이터의 정의


Ex). 시험점수 85점, 주문 금액 5만원, 기온 27도 등
     모두 관찰 및 측정된 데이터


- 값 자체는 사실을 담지만 의미는 맥락에 따라 변화.
- 85점의 높낮음은 난이도 및 평균, 이전 성적을 함께 봐야 판단 가능
- 데이터는 결론이 아니라 질문에 답하기 위해 해석할 원재료

행(row) = record

## 데이터와 정보
85점에 학기 평균보다 10점 높음이란 비교를 더하면 의미 있는 정보로 변화

- 데이터 : 가공 전의 사실과 값
- 정보 : 목적에 맞게 정리하고 맥락을 부여한 결과
- 같은 데이터도 질문과 가공 방법에 따라 다른 정보가 될 수 있음

## 정보와 통찰
분석의 가치는 무엇을 해야 하는지 판단할 수 있을 때 완성

- 데이터(data) : 매출이 15% 감소
- 정보(information) : 방문자는 유지 / 모바일 결제 이탈자 상승
- 통찰(insight) : 모바일 결제 화면 우선 개선 효과 측정 
반정형이 정형보다 유동적 

## 데이터 구분의 필요성
두 관점을 분리해야 숫자로 저장된 고객 번호를 측정값으로 오해 안 함

- 형태 관점: 정형/반정형/비정형 
- 성격 관점: 수치형/범주형

저장 형식과 현실 의미의 동시 확인을 통한 적절한 분석 방법 선택


## 형태에 따른 데이터 분류
데이터 형태는 파일 확장자가 아니라 항목의 위치/의미/일관성에 관한 구조적 특성 <br>

- 정형 : 행과 열, 스키마가 명확
- 반정형 : 키와 태그로 의미를 표현하며 구조가 유연함
- 비정형 : 분석에 바로 쓸 고정 구조가 없어 특징 추출 필요 

## 정형 데이터(structured data)
각 값이 들어갈 위치와 의미가 명확 <br>
같은 열에는 같은 형식의 값이 반복 <br>
**값의 위치와 의미가 일정해 선택/비교/집계 안정적 수행**

- 대표 사례 : 관계형 DB, Excel표, CSV <br>
  행은 하나의 관측 대상, 열은 공통 속성을 나타냄
  표처럼 보이는 것보다 행.열의 의미가 일관적인지 중요 


### 정형 데이터의 행과 열 
한 행은 한 고객의 레코드 한 열은 모든 고객에게 공통으로 적용되는 정보의 종류 



고객 번호는 기록을 구분하는 식별값 <br>
나이는 수치 속성, 회원 등급은 범주 속성<br>
**행과 열의 역할이 명확해야 대상과 속성을 선택 비교 가능**

## 스키마
**컴퓨터가 각 값을 어떻게 처리할 지 알려주는 정형 데이터의 설게도**

자료형 : 나이는 정수, 이름은 문자열, 금액은 숫자로 정의<br>
제약 조건 : 필수. 고유 고객 번호 설정<br>
스키마가 있어도 값의 실제 의미/단위/품질은 별도 검증 필요

## 반정형 데이터 
**고정된 행과 열 대신 키/태그/구분자가 각 값의 의미와 관계 알려줌**

- 대표 사례 : JSON/XML/HTML 등


### 반정형 데이터의 분석 준비
**중첩 객체와 배열을 풀고 서로 다른 항목을 하나의 분석용 표로 정리**
- 중첩 해제 : 객체를 열, 배열을 여러 행으로 변환
- 자료형 통일 : 숫자 2와 문자열 2를 같은 기준으로 변환
- 항목 통일 : 없는 필드는 결측치로 처리, 키이름을 표준화 

## 비정형 데이터
**자연어/이미지/음성/영상은 사람이 이해할 수 있어도 컴퓨터 분석에서는 별도의 특징 추출 필요**

- 텍스트 
- 시각 음향
