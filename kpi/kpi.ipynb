# KPI (key performance indicator)
:  핵심성과지표 <br>
    조직, 부서, 팀 또는 개인이 설정한 목표를 어느 정도 달성하고 있는지 판단하기 위해 선택한 핵심 측정값 

- 달성하려는 목표
- 현재 상태와 목표의 차이
- 성과의 개선 또는 악화 추세
- 차이가 발생한 영역
- 현재 결과에 대응할 행동
- 실행한 조치의 효과 
  

key : 조직이 반드시 집중해야하는 소수의 지표 <br>




데이터 셋 : 데이터들이 모여있는 상태 



데이터는 사실과  값을 기록 <br>
메트릭은 기록된 데이터를 가공하여 측정 <br>
KPI는 목표의 달성 여부를 판단 (핵심성과지표) 

데이터 < 메트릭 < KPI  
 -  KPI는 메트릭이 될 수 있지만 메트릭은 KPI가 될 수 없다.
 - 동일 개념으로 메트릭은 데이터가 될 수 있지만 메트릭이 될 수 없다.
 - 상위 개념은 하위개념일 수 있지만 하위개념은 상위개념이 될 수 없다. 

## 메트릭이 KPI 되는 방법

- 목표와 직접적으로 연결 되어 있는지 여부 확인
- 결과에 따라 구체적 행동을 결정 할 수 있는지 확인
- 측정 대상과 기간의 명확 여부 확인
- 목표값이나 판단 기준을 설정 가능 여부 확인
- 지속적 측정 가능 여부 확인

## 절대값 / 비율 / 변화율 

### 절대값 

- 특정 기간이나 시점에 발생한 대상의 수량이나 규모를 그대로 나타낸 값 
  - ex). 매출액 10억/고객 2천명 / 주문 5천건 / 불량 12건 
  
but 절대값으로만 공정 비교 어려움 <br>
 A = 생산량 1천개 / 불량품 20개 -> 불량률 2%<br>
 B = 생산량 10만개 / 불량품 100개 -> 불량률 0.1%

- 품질 수준을 비교하려면 불량 갯수가 아닌 전체 생산량 대비 불량 비율 확인해야 함

### 비율

- 특정 값이 전체나 기준값에서 차지하는 상대적 크기 표현
- 일반적으로 분자 분모 관계로 계산 / 백분율, 소수 형태로 표현 
  

### 변화율 

- 기준 시점의 값과 비교 시점의 값 사이에 얼마나 변화가 발생했는지 상대적 표현
- 일반적으로 비교 시점 값 - 기준 시점 값 후 차이를 기준 시점 값으로 나눠 계산
- 매출이 100에서 120 증가하면 증가량은 20이고 증가율은 20%
- 매출이 20에서 40으로 증가하면 증가량은 20이고 증가율은 100%
  
즉 비율의 변화에서는 퍼센트와 퍼센트포인트를 구분해야함<br>

ex). 10% -> 12% 퍼센트로만 따지면 2% 상승<br>
 증가율로 치면 20% 상승이다. 

## 좋은 KPI 조건

:SMART 원칙이 널리 활용됨

### SMART 

- 목표를 명확하고 실행 가능한 형태로 정의하기 위해 사용되는 대표적인 기준
  - Speicific
  - Measurable
  - Achievable
  - Relevant
  - Time-bound
  
- SMART 적용 시 수치 외 정의.행동.책임까지 검토
- SMART의 한계 : 자동 KPI 생성 공식과 구분
- SMART는 시작점이고 뒤의 실무 검토가 함께 필요함.

#### 구체성(SPECIFIC)
- 목표와 KPI의 측정 대상, 범위, 계산 기준이 분명해야 한다는 의미 
  - 고객 정의에서도 동일한 경계 설정 필요
  - 구체성은 문장을 길게 쓰는 것이 아니라 경계를 명확하게 만드는 일
  - 일관된 계산을 위한 한 문장 정의 필요 

#### 측정 가능성(Measurable)
- 계산 가능성보다 포괄적인 개념
  - 데이터 존재와 측정 가능성의 구분
  - 비율에서는 분자와 분모가 같은 모집단과 기간을 확보
  - 좋은 측정은 계산 가능성뿐 아니라 재현성과 대표성을 포함

#### 달성 가능성(Achievable)
- 현재 조건과 자원을 고려할 때 목표가 현실적인 도전 범위에 있어야 한다는 의미
  - 달성 가능성과 쉬운 목표의 구분 
    - 최근 3년 성장률이 5%였던 조직이 변화 없이 다음 해 50%성장 요구 -> 실행 근거 부족
  - 분석가는 목표 숫자만 평가하기 x 숫자를 만들 변화 가설과 자원까지 확인
  

#### 관련성(Relevant)
- KPI가 조직의 목표와 전략에 실제로 연결되어 있어야 한다는 의미
  - 유료 매출 성장을 위한 방문자 수 외 전환.수익성 지표 필요
  - 관련성은 조직의 전략에 따라 변화
  - 다른 회사의 KPI를 그대로 복사하기 전에 우리의 목표와 인과 경로를 먼저 확인

#### 기한과 측정 주기(Time-bound)
- 언제까지 목표를 달성할지와 얼마나 자주 측정할지가 명확해야 한다는 의미
  - 측정 주기는 업무의 변화 속도와 대응 가능성에 일치
  - 주기가 지나치게 짧으면 적은 표본의 자연스러운 흔들림에 과민 반응 / 너무 짧으면 문제 발견이 지연
  - 목표 기간은 방향 / 측정 주기는 확인 간격으로 이해 
  

## 주의가 필요한 지표 

### 허영지표
- 누적 가입자 수는 혀영지표가 되기 쉬운 대표 사례
  - 목적에 따라 허영지표로 변하는 활동량 자표
  - 허영지표를 개선하려면 행동과 결과에 가까운 지표를 함께 확인
  - 큰 숫자의 패기가 아닌 올바른 맥락 배치 <br><br>


### 대리지표
- 고객 신뢰, 직원 역량, 제품 가치처럼 직접 측정하기 어려운 개념을 대신 관찰하는 지표
  - 대리값과 원래 목표의 관계 미검증에 따른 문제 발생
  - 대리지표를 사용할 때에는 측정하려는 개념을 분명히 밝히고, 결과지표와 관계를 주기적 확인
  - 현실과 대리지표의 명확한 구분
  <br><br>


## KPI 검토 질문
- 환경 변화에 따른 KPI 재검토 필요
  - 목표와 관련 있는가
  - 정의와 기간이 구체적인가
  - 데이터를 반복 측정할 수 있는가
  - 담당자가 영향을 줄 수 있는가
  - 결과에 따라 행동할 수 있는가
  - 측정 비용이 합리적인가
  - 부작용을 막을 지표가 있는가 
- 전략과 고객, 데이터 수집 방식이 바뀌면 지표의미도 변화 
<br><br>



## MECE 분류 원칙 ( Mutually Exclusive / Collectively Exhaustive)
- 상호 배타적이고 전체 포괄적인 분류를 의미 
  - 고객을 신규와 기존으로 나눈다 -> 신규의 기간 기준을 정해 고객이 동시에 두 집단에 중복 소속을 방지
  - 현실을 완벽히 설명하는 철학보단 집계 오류를 줄이는 점검 원칙
  - KPI 트리에서는 하위 합계가 상위 값과 일치하는지 우선 검증 필요 <br><br>



## GQM 접근 ( Goal - Question - Metric)
- 목표를 정의, 목표달성 여부 판단 질문 생성, 질문에 답할 메트릭 선택 방법
  - 질문에 답하지 못하는 지표는 제외
  - 지표부터 고르는 습관을 막고 측정 목적과 데이터 사이의 논리 표출
