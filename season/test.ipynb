```python
import pandas as pd
from sklearn.linear_model import LinearRegression

df = pd.DataFrame({
    '기온': [20,22,24,26,28,30],
    '판매량': [100, 120, 140, 160, 180, 200]
})

X = df[['기온']]
y = df['판매량']

model = LinearRegression()
model.fit(X, y)

q = pd.DataFrame({'기온': [27]})
prediction = model.predict(q)

print(f'{prediction[0]:.0f}개')
```

```python
from sklearn.cluster import KMeans

df = pd.DataFrame({
    '고객': ['A','B', 'C','D','E','F'],
    '방문횟수': [2,3,2,18,20,19],
    '구매횟수': [1,2,2,15,17,16]
})

X = df[['방문횟수', '구매횟수']]

model = KMeans(n_clusters=2, random_state=42, n_init=10)

df['그룹'] = model.fit_predict(X)

df
```
