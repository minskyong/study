```python
dtype_series = pd.Series([1,2,3,4], dtype = 'float')    #datatype 변경 가능 dtype

d_s = dtype_series

d_s
```

```python
float_series = pd.Series([64.0, 75.2, 80.0],
                dtype ='float64'   #float -> int 변경시 데이터 손실 발생 ex. 75.2 --> 75

)

float_series.astype('int64')
```

```python
#같은 인덱스가 있을 때
a = pd.Series([1,2,3,], index=['x','y','z'])            #np배열있어서 같은 인덱스끼리 연산 수행 
b = pd.Series([10,20,30], index=['z','y','x'])
a+b
```

```python
#다른 인덱스가 있을 때
c= pd.Series([1,2], index= ['x','y'])         #짝 없는 데이터 버리거나 NAN처리 
d= pd.Series([10,20], index =['y','z'])       #NAN은 실수형으로 적용 
c+d
```

```python
#dic랑 같은구조 = 키 : 값   
#시리즈에 딕셔너리 넣기 가능 
name = {
        '김' : 2,   
        '이' : 3,
        '박' : 4
}


pd.Series(name)
```

```python
data = { 
        '서울' : 10,
        '부산' : 20,
        '대전' : 30,
        '대구' : 40
}

labeled_data = pd.Series(data, index = ['서울','부산','광주'])

labeled_data
```
