```python
import pandas as pd
import numpy as np

orders = pd.DataFrame(
    {
        "order_date": pd.to_datetime([
            "2026-08-01", "2026-08-01", "2026-08-02", "2026-08-03",
            "2026-08-04", "2026-08-05", "2026-08-05", "2026-08-06",
            "2026-08-07", "2026-08-08", "2026-08-09", "2026-08-10",
        ]),
        "customer": pd.Series(
            ["민준", "서연", "지우", "도윤", "하준", "서연",
             "민준", "지우", "도윤", "하준", "민준", "서연"],
            dtype="string",
        ),
        "region": pd.Series(
            ["서울", "부산", "서울", "대전", "부산", "서울",
             "대전", "서울", "부산", "대전", "서울", "부산"],
            dtype="category",
        ),
        "product": pd.Series(
            ["키보드", "모니터", "마우스", "키보드", "모니터", "키보드",
             "마우스", "모니터", "키보드", "마우스", "모니터", "키보드"],
            dtype="category",
        ),
        "quantity": pd.Series(
            [2, 1, 3, 2, 1, 4, 2, 1, 5, 2, 1, 3],
            dtype="Int64",
        ),
        "unit_price": [
            45000, 320000, 25000, 45000, 320000, 45000,
            25000, 320000, 45000, 25000, 320000, 45000,
        ],
        "discount_rate": [
            0.00, 0.10, np.nan, 0.05, 0.00, 0.10,
            0.00, np.nan, 0.05, 0.00, 0.10, 0.00,
        ],
        "paid": pd.Series(
            [True, True, False, True, True, True,
             False, True, True, pd.NA, True, False],
            dtype="boolean",
        ),
    },
    index=pd.Index(
        [f"O{i:03d}" for i in range(1, 13)],
        name="order_id",
    ),
)

orders["customer"] = pd.array(
    ["민준", "서연", "지우", "도윤", "하준", "서연",
     "민준", "지우", "도윤", "하준", "민준", "서연"],
    dtype="string",
)
orders["region"] = pd.Categorical(
    ["서울", "부산", "서울", "대전", "부산", "서울",
     "대전", "서울", "부산", "대전", "서울", "부산"]
)
orders["product"] = pd.Categorical(
    ["키보드", "모니터", "마우스", "키보드", "모니터", "키보드",
     "마우스", "모니터", "키보드", "마우스", "모니터", "키보드"]
)
orders["quantity"] = pd.array(
    [2, 1, 3, 2, 1, 4, 2, 1, 5, 2, 1, 3],
    dtype="Int64",
)
orders["paid"] = pd.array(
    [True, True, False, True, True, True,
     False, True, True, pd.NA, True, False],
    dtype="boolean",
)

print(orders)
```
